```python
import random

def determine_label(sample):
    first_half = sample[:3]
    second_half = sample[3:]

    if any(val in (1, 2) for val in first_half):
        return 1
    elif any(val in (3, 4) for val in first_half):
        return 3
    elif any(val in (1, 2) for val in second_half):
        return 2
    elif any(val in (3, 4) for val in second_half):
        return 4
    else:
        return 0  # All values in 5,6,7

def is_valid_sample(sample):
    first_half = sample[:3]
    second_half = sample[3:]

    first_has_1_to_4 = any(val in (1, 2, 3, 4) for val in first_half)
    second_has_1_to_4 = any(val in (1, 2, 3, 4) for val in second_half)

    # Enforce exclusivity: if one half has low values, the other half must be high only (5-7)
    if first_has_1_to_4 and not all(val in (5, 6, 7) for val in second_half):
        return False
    if second_has_1_to_4 and not all(val in (5, 6, 7) for val in first_half):
        return False

    return True

def generate_balanced_dataset(n_total):
    assert n_total % 5 == 0, "n_total must be divisible by 5."
    n_per_class = n_total // 5

    seen = set()
    data = []
    labels = []
    counts = [0] * 5

    while sum(counts) < n_total:
        sample = tuple(random.randint(1, 7) for _ in range(6))  # values now in 1–7

        if sample in seen or not is_valid_sample(sample):
            continue

        label = determine_label(sample)

        if counts[label] < n_per_class:
            seen.add(sample)
            data.append(list(sample))
            labels.append(label)
            counts[label] += 1

    return [data, labels]
```
