```python
import random

def determine_label(sample):
    if any(val in (1, 2, 3) for val in sample):
        return 1
    else:
        return 0

def generate_balanced_dataset(n_total):
    assert n_total % 2 == 0, "n_total must be divisible by 2."
    n_per_class = n_total // 2

    seen = set()
    data = []
    labels = []
    counts = [0, 0]  # For label 0 and 1

    while sum(counts) < n_total:
        sample = tuple(random.randint(1, 6) for _ in range(6))
        if sample in seen:
            continue

        label = determine_label(sample)

        if counts[label] < n_per_class:
            seen.add(sample)
            data.append(list(sample))
            labels.append(label)
            counts[label] += 1

    return [data, labels]

# Example usage
if __name__ == "__main__":
    DatasTrain = generate_balanced_dataset(100)

```
