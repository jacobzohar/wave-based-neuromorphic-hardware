```python
import random

def determine_label(sample):
    first_half = sample[:3]
    second_half = sample[3:]

    if any(val in (1, 2, 3) for val in first_half):
        return 1
    elif any(val in (1, 2, 3) for val in second_half):
        return 2
    else:
        return 0  # All values are in 4–6

def is_valid_sample(sample):
    first_half = sample[:3]
    second_half = sample[3:]

    first_has_1_to_3 = any(val in (1, 2, 3) for val in first_half)
    second_has_1_to_3 = any(val in (1, 2, 3) for val in second_half)

    # Exclusivity enforcement:
    if first_has_1_to_3 and not all(val in (4, 5, 6) for val in second_half):
        return False
    if second_has_1_to_3 and not all(val in (4, 5, 6) for val in first_half):
        return False

    return True

def generate_balanced_dataset(n_total):
    assert n_total % 3 == 0, "n_total must be divisible by 3."
    n_per_class = n_total // 3

    seen = set()
    data = []
    labels = []
    counts = [0] * 3

    while sum(counts) < n_total:
        sample = tuple(random.randint(1, 6) for _ in range(6))

        if sample in seen or not is_valid_sample(sample):
            continue

        label = determine_label(sample)

        if counts[label] < n_per_class:
            seen.add(sample)
            data.append(list(sample))
            labels.append(label)
            counts[label] += 1

    return [data, labels]

# Example usage
if __name__ == "__main__":
    DatasTrain = generate_balanced_dataset(2000)
```
